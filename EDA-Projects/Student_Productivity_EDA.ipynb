```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='study_hours_per_day',
    y='productivity_score'
)

plt.title('Study Hours vs Productivity Score')
plt.xlabel('Study Hours per Day')
plt.ylabel('Productivity Score')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='phone_usage_hours',
    y='productivity_score'
)

plt.title('Phone Usage vs Productivity Score')
plt.xlabel('Phone Usage Hours per Day')
plt.ylabel('Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='sleep_hours',
    y='productivity_score'
)

plt.title('Sleep Hours vs Productivity Score')
plt.xlabel('Sleep Hours per Day')
plt.ylabel('Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='social_media_hours',
    y='productivity_score'
)

plt.title('Social Media Usage vs Productivity Score')
plt.xlabel('Social Media Hours per Day')
plt.ylabel('Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(14, 10))

corr = df.drop(columns=['student_id']).corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Student Productivity')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='stress_level',
    y='productivity_score'
)

plt.title('Stress Level vs Productivity Score')
plt.xlabel('Stress Level')
plt.ylabel('Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df['productivity_score'])

plt.title('Box Plot of Productivity Score')
plt.ylabel('Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x='gender',
    y='productivity_score',
    errorbar=None
)

plt.title('Average Productivity Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Productivity Score')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='productivity_score',
    bins=20,
    kde=True
)

plt.title('Distribution of Productivity Scores')
plt.xlabel('Productivity Score')
plt.ylabel('Number of Students')

plt.show()
```

Conclusion


The Exploratory Data Analysis of the Student Productivity and Digital Distraction dataset provided insights into the relationships between student habits and productivity scores.

The correlation analysis showed a strong positive relationship between study hours and productivity score (0.733). Focus score (0.411) and sleep hours (0.341) also showed positive relationships. Phone usage hours (-0.327) and stress level (-0.197) showed negative relationships with productivity score.

The visualizations helped identify patterns in student productivity, compare average scores across gender categories, examine score distributions, and identify potential outliers.

Overall, the analysis suggests that study habits, focus, sleep, phone usage, and stress are associated with productivity scores in this dataset. However, correlation does not establish causation.
